```python
import psycopg2
from psycopg2 import sql
from psycopg2.extras import execute_values, Json
import logging
import json
import html
import re
import csv
import os
from tqdm.notebook import tqdm
```

```python
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')
```

```python
DB_PARAMS = {
    'dbname':   'thecall',
    'user':     'postgres',
    'password': 'password',
    'host':     'localhost',
    'port':     '5432'
}

TABLE_SCHEMA = 'articles'
TABLE_NAME   = 'entities'

# Output directory for CSV files
OUTPUT_DIR = '.'
```

```python
# ── Database helpers ──────────────────────────────────────────────────────────

def connect_to_db(params):
    try:
        logging.info("Connecting to PostgreSQL...")
        conn = psycopg2.connect(**params)
        conn.autocommit = False
        logging.info("Connected.")
        return conn
    except psycopg2.Error as e:
        logging.error(f"Connection failed: {e}")
        return None
```

```python
# ── Statistics helpers ────────────────────────────────────────────────────────

def get_type_distribution(cursor, schema, where_clause='', params=None):
    """Return entity-type counts and percentages for the given filter."""
    query = f"""
        SELECT
            entity_type,
            COUNT(*)                                      AS cnt,
            ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) AS pct
        FROM {schema}.entities
        {where_clause}
        GROUP BY entity_type
        ORDER BY cnt DESC
    """
    cursor.execute(query, params)
    return cursor.fetchall()


def get_top_entities_by_type(cursor, schema, entity_type, top_n=20,
                             where_clause='', params=None):
    """Return the top N most frequent entity values for a given type."""
    base_where = f"WHERE entity_type = %s"
    if where_clause:
        # where_clause already starts with WHERE; convert first one to AND
        combined = where_clause.replace('WHERE', 'AND', 1)
        full_where = base_where + ' ' + combined
        combined_params = (entity_type,) + (params or ())
    else:
        full_where = base_where
        combined_params = (entity_type,)

    query = f"""
        SELECT entity_value, COUNT(*) AS cnt
        FROM {schema}.entities
        {full_where}
        GROUP BY entity_value
        ORDER BY cnt DESC
        LIMIT %s
    """
    cursor.execute(query, combined_params + (top_n,))
    return cursor.fetchall()


def get_all_entity_counts(cursor, schema, where_clause='', params=None):
    """Return every (entity_type, entity_value, count) row for CSV export."""
    query = f"""
        SELECT entity_type, entity_value, COUNT(*) AS cnt
        FROM {schema}.entities
        {where_clause}
        GROUP BY entity_type, entity_value
        ORDER BY entity_type, cnt DESC, entity_value
    """
    cursor.execute(query, params)
    return cursor.fetchall()


def write_entity_csv(rows, filepath):
    """Write (entity_type, entity_value, count) rows to a CSV file."""
    with open(filepath, 'w', newline='', encoding='utf-8') as f:
        writer = csv.writer(f)
        writer.writerow(['entity_type', 'entity_value', 'count'])
        writer.writerows(rows)
    logging.info(f"Wrote {len(rows):,} rows to {filepath}")


def print_type_distribution(rows, label):
    print(f"\n{'='*60}")
    print(f"  Entity-type distribution — {label}")
    print(f"{'='*60}")
    print(f"{'Type':<10} {'Count':>12} {'Percentage':>12}")
    print('-' * 36)
    total = sum(r[1] for r in rows)
    for entity_type, cnt, pct in rows:
        print(f"{entity_type:<10} {cnt:>12,} {pct:>11.2f}%")
    print('-' * 36)
    print(f"{'TOTAL':<10} {total:>12,} {'100.00':>11}%")


def print_top_entities(top_n_results, label):
    print(f"\n{'='*60}")
    print(f"  Most common entities — {label}")
    print(f"{'='*60}")
    for entity_type, rows in top_n_results.items():
        print(f"\n  [{entity_type}]")
        print(f"  {'Value':<45} {'Count':>8}")
        print('  ' + '-' * 55)
        for value, cnt in rows:
            print(f"  {str(value):<45} {cnt:>8,}")
```

```python
# ── Analysis runner ───────────────────────────────────────────────────────────

def run_analysis(cursor, schema, label, csv_path,
                 where_clause='', params=None, top_n=20):
    """
    Run the full suite of descriptive statistics for a given filter
    and write results to csv_path.
    """
    logging.info(f"Running analysis: {label}")

    # 1. Type distribution
    dist = get_type_distribution(cursor, schema, where_clause, params)
    print_type_distribution(dist, label)

    # 2. Top N per type
    entity_types = ['PER', 'MISC', 'ORG', 'LOC']
    top_n_results = {}
    for et in entity_types:
        top_n_results[et] = get_top_entities_by_type(
            cursor, schema, et, top_n, where_clause, params
        )
    print_top_entities(top_n_results, label)

    # 3. Full entity-value count CSV
    all_rows = get_all_entity_counts(cursor, schema, where_clause, params)
    write_entity_csv(all_rows, csv_path)
```

```python
# ── Main ──────────────────────────────────────────────────────────────────────

def main():
    conn = connect_to_db(DB_PARAMS)
    if not conn:
        return

    try:
        cursor = conn.cursor()

        schema = TABLE_SCHEMA  # 'articles'

        # ── Dataset A: full entities table ────────────────────────────────────
        csv_all = os.path.join(OUTPUT_DIR, 'entity_counts_all.csv')
        run_analysis(
            cursor, schema,
            label       = 'ALL entities',
            csv_path    = csv_all,
            where_clause = '',
            params       = None,
            top_n        = 20
        )

        # ── Dataset B: entities linked to nitearticle = 1 rows ────────────────
        # Join entities -> filelist on fk_filelist = filelist.id
        # and keep only rows where filelist.nitearticle = 1
        nite_where = """
            WHERE fk_filelist IN (
                SELECT id FROM articles.filelist WHERE nitearticle = 1
            )
        """
        csv_nite = os.path.join(OUTPUT_DIR, 'entity_counts_nitearticle.csv')
        run_analysis(
            cursor, schema,
            label        = 'entities where nitearticle = 1',
            csv_path     = csv_nite,
            where_clause = nite_where,
            params       = None,
            top_n        = 20
        )

        cursor.close()

    finally:
        conn.close()
        logging.info("Database connection closed.")
        logging.info("Done.")
```

```python
if __name__ == '__main__':
    main()
```
