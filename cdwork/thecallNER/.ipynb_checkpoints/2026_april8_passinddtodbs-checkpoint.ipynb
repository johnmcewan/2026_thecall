```python
import psycopg2
from psycopg2 import sql, extras
import logging
import csv
import os
import json
```

```python
# Set up basic logging to see success/failure messages
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')
```

```python
DB_PARAMS = {
    'dbname': 'thecall',
    'user': 'postgres',        
    'password': 'password',    
    'host': 'localhost',           
    'port': '5432'                 
}
```

```python
def connect_to_db(params):
    """Establishes a connection to the PostgreSQL database."""
    conn = None
    try:
        logging.info("Attempting to connect to PostgreSQL database...")
        conn = psycopg2.connect(**params)
        # Set autocommit to False so we can control transactions (needed for UPDATE)
        conn.autocommit = False 
        logging.info("Connection successful.")
        return conn
    except psycopg2.Error as e:
        logging.error(f"Error connecting to the database: {e}")
        return None
```

```python
def collect_data_from_db(conn, table_schema, table_name):
    full_table_name = f"{table_schema}.{table_name}"
    try:
        with conn.cursor() as cur:
            # Fetch filename and folder to match with local files
            query = sql.SQL("SELECT filename, folder FROM {}.{} WHERE json_raw is NULL").format(
                sql.Identifier(table_schema),
                sql.Identifier(table_name)
            )
            cur.execute(query)
            return cur.fetchall()
    except psycopg2.Error as e:
        logging.error(f"Error fetching data: {e}")
        return []
```

```python
def matching_task(db_rows):
    root_dir = r"E:\Callproject\extracted_json"
    results = []
    
    # DB rows are (filename, folder). 
    # Let's normalize them into a set of (folder\filename) for fast lookup
    # We convert both to lowercase to avoid case-sensitivity issues
    db_lookup = {os.path.join(row[1], row[0]).lower(): (row[0], row[1]) for row in db_rows}
    
    logging.info(f"Scanning directory: {root_dir}")
    
    for root, dirs, files in os.walk(root_dir):
        for file in files:
            if file.endswith('.json'):
                # 1. Get the path relative to the 'extracted_json' folder
                # If your DB 'folder' starts with 'E:\Callproject\assembled', 
                # we need to map the 'extracted_json' path back to that structure.
                
                # Replace the 'extracted_json' part of the path to match the DB 'assembled' path
                current_folder = root.replace("extracted_json", "assembled")
                
                # 2. Change extension from .json to .indd to match the DB filename
                filename_in_db = file.replace(".json", ".indd")
                
                # 3. Construct the full path key for lookup
                full_path_key = os.path.join(current_folder, filename_in_db).lower()
                
                if full_path_key in db_lookup:
                    # Get the original casing from the DB for the UPDATE criteria
                    orig_filename, orig_folder = db_lookup[full_path_key]
                    
                    file_path = os.path.join(root, file)
                    try:
                        with open(file_path, 'r', encoding='utf-8') as f:
                            json_data = json.load(f)
                            results.append((json.dumps(json_data), orig_filename, orig_folder))
                    except Exception as e:
                        logging.error(f"Failed to read {file}: {e}")

    logging.info(f"Matched {len(results)} files to database rows.")
    return results
```

```python
def update_dbs(conn, table_schema, table_name, results):
    if not results:
        logging.info("No results to update.")
        return

    try:
        with conn.cursor() as cur:
            # We add ::json to t.json_content to cast the text to a JSON type
            update_query = """
                UPDATE articles.filelist AS f
                SET json_raw = t.json_content::json
                FROM (VALUES %s) AS t(json_content, filename, folder)
                WHERE f.filename = t.filename AND f.folder = t.folder
            """
            
            extras.execute_values(cur, update_query, results)
            conn.commit()
            logging.info(f"Successfully updated {len(results)} rows in the database.")
            
    except psycopg2.Error as e:
        conn.rollback()
        logging.error(f"Database update failed: {e}")
```

```python
def main():
    conn = connect_to_db(DB_PARAMS)
    if conn:
        try:
            table_schema = 'articles'
            table_name = 'filelist'

            # 1. Fetch existing entries
            all_data = collect_data_from_db(conn, table_schema, table_name)
            
            if not all_data:
                logging.warning("No data retrieved from DB. Exiting.")
                return

            # 2. Match local JSONs with DB entries
            results = matching_task(all_data)

            # 3. Batch update the DB
            update_dbs(conn, table_schema, table_name, results)
            
        finally:
            if conn:
                conn.close()
                logging.info("Database connection closed.")
```

```python
if __name__ == '__main__':
    main()
```
