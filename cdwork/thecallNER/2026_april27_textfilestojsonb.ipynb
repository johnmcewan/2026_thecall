```python
import json
import os

def text_to_unified_json(file_path, db_id, serial="000000"):
    """
    Converts a plain text file into the unified JSON format 
    used for QuarkXPress and InDesign extractions.
    """
    # Read text and split into paragraphs (skipping empty lines)
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = [line.strip() for line in f.readlines() if line.strip()]
    
    filename = os.path.basename(file_path)
    
    # 1. Build the paragraph array
    paragraphs = [{"text": line, "style": None} for line in lines]
    
    # 2. Build the HTML representation
    html_paragraphs = "\n".join([f'    <p class="ad-para">{line}</p>' for line in lines])
    text_html = f"""<article class="ad-card" data-format="text" data-id="{db_id}" data-serial="{serial}">
  <h2 class="ad-title">{filename}</h2>
  <div class="ad-frame" data-frame-id="1">
{html_paragraphs}
  </div>
</article>"""
    
    # 3. Build plain text
    text_plain = filename + "\n" + "\n".join(lines)
    
    # 4. Construct final payload
    unified_data = {
        "db_id": db_id,
        "frames": [{
            "text": " ".join(lines),
            "bounds": None,
            "frame_id": "1",
            "paragraphs": paragraphs
        }],
        "serial": serial,
        "text_html": text_html,
        "text_plain": text_plain,
        "layout_name": filename,
        "source_file": file_path,
        "source_format": "text"
    }
    
    return unified_data
```
