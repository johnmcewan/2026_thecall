```python
import subprocess
import os
import json
from pathlib import Path
import pdfplumber
import shutil
from pathlib import Path
import re
```

```python
import psycopg2
from psycopg2 import sql, extras
import logging
```

```python
from psycopg2.extras import RealDictCursor
```

```python
# Set up basic logging to see success/failure messages
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')
```

```python
DB_PARAMS = {
    'dbname': 'thecall',
    'user': 'postgres',        
    'password': 'password',    
    'host': 'localhost',           
    'port': '5432'                 
}
```

```python
def connect_to_db(params):
    """Establishes a connection to the PostgreSQL database."""
    conn = None
    try:
        logging.info("Attempting to connect to PostgreSQL database...")
        conn = psycopg2.connect(**params)
        # Set autocommit to False so we can control transactions (needed for UPDATE)
        conn.autocommit = False 
        logging.info("Connection successful.")
        return conn
    except psycopg2.Error as e:
        logging.error(f"Error connecting to the database: {e}")
        return None
```

```python
def collect_data_from_db(conn, table_schema, table_name):
    try:
        with conn.cursor(cursor_factory=RealDictCursor) as cur:
            query = sql.SQL("SELECT * FROM {}.{} WHERE nitearticle = 1").format(
                sql.Identifier(table_schema),
                sql.Identifier(table_name)
            )
            cur.execute(query)
            return cur.fetchall()  # This now returns a list of dictionaries
    except psycopg2.Error as e:
        logging.error(f"Error fetching data: {e}")
        return []
```

```python
def extract_to_standard_json(all_data, output_folder):
    """
    all_data: List of records fetched from Postgres (e.g., list of dicts or tuples)
    output_folder: Path where the JSON file should be saved
    """
    results = []
    
    logging.info(f"Preparing to extract {len(all_data)} records to JSON...")

    # Iterate through the data retrieved from the database
    for row in all_data:
        # Assuming all_data is a list of dictionaries (standard for RealDictCursor)
        # If it's a list of tuples, you would map them: row[0], row[1], etc.
        article_entry = {
            "id": row.get('id'),
            "source_file": row.get('filename'),
            "filedate": row.get('filedate'),
            "folder": row.get('folder'),
            "text": row.get('json_raw'),  # Or whatever your content column is named          
        }
        results.append(article_entry)

    # Ensure the output directory exists
    os.makedirs(output_folder, exist_ok=True)
    
    output_path = os.path.join(output_folder, "MNF_articles.json")

    try:
        with open(output_path, 'w', encoding='utf-8') as f:
            # Use indent=4 for a human-readable file, or None for a minified version
            # default=str handles UUIDs or Datetimes from Postgres
            json.dump(results, f, indent=4, ensure_ascii=False, default=str)
            
        logging.info(f"Successfully saved {len(results)} articles to {output_path}")
    except Exception as e:
        logging.error(f"Failed to write JSON file: {e}")
        return None

    return output_path
```

```python
def main():
    conn = connect_to_db(DB_PARAMS)

    OUTPUT_FOLDER = r"D:/development/2026_python/2026_thecall/website6"
 
    if conn:
        try:
            table_schema = 'articles'
            table_name = 'filelist'

            # 1. Fetch existing entries
            all_data = collect_data_from_db(conn, table_schema, table_name)
            
            if not all_data:
                logging.warning("No data retrieved from DB. Exiting.")
                return

        # 2. Process files and generate JSON strings
            results = extract_to_standard_json(
                all_data, 
                OUTPUT_FOLDER, 
            )
                        
        finally:
            if conn:
                conn.close()
                logging.info("Database connection closed.")
```

```python
if __name__ == "__main__":
    main()
```
