```python
import subprocess
import os
import shutil
from pathlib import Path
```

```python
def extract_quark_text_fixed(source_dir, target_dir, soffice_path):
    source_root = Path(source_dir).resolve()
    target_root = Path(target_dir).resolve()
    
    # Clean the environment to prevent library prefix/conflict errors
    # This prevents the "Could not find platform independent libraries" message
    clean_env = os.environ.copy()
    clean_env.pop("PYTHONHOME", None)
    clean_env.pop("PYTHONPATH", None)
    
    # Create a local temp directory for renaming files
    temp_processing_dir = target_root / "_temp_ext_injection"
    temp_processing_dir.mkdir(parents=True, exist_ok=True)

    print(f"Scanning: {source_root}")

    for file_path in source_root.rglob("*"):
        # Skip directories and the temp processing folder itself
        if file_path.is_dir() or "_temp_ext_injection" in str(file_path):
            continue

        # Step 1: Force extension injection. 
        # LibreOffice often fails to load 'MMXPR3' binaries if they don't end in .qxd
        temp_file = temp_processing_dir / f"{file_path.name}.qxd"
        shutil.copy2(file_path, temp_file)
        
        # Determine output location (mirroring source structure)
        relative_path = file_path.relative_to(source_root)
        output_subfolder = target_root / relative_path.parent
        output_subfolder.mkdir(parents=True, exist_ok=True)

        print(f"Processing: {file_path.name}")

        try:
            # Step 2: Use 'txt:Text' for maximum stability with libmwaw
            subprocess.run([
                soffice_path,
                "--headless",
                "--convert-to", "txt:Text",
                "--outdir", str(output_subfolder),
                str(temp_file)
            ], check=True, capture_output=True, env=clean_env)
            
            # Remove temp file immediately after successful call
            temp_file.unlink()

        except subprocess.CalledProcessError as e:
            print(f"!! Error on {file_path.name}: {e.stderr.decode().strip()}")
        except Exception as e:
            print(f"!! Unexpected error: {e}")

    # Final cleanup
    if temp_processing_dir.exists():
        shutil.rmtree(temp_processing_dir)
    print(f"\nTask Finished. Check results in: {target_root}")
```
