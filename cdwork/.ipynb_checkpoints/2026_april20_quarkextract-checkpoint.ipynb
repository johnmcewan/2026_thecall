```python
"""
batch_quark_export.py
Batch-converts QuarkXPress _9x.qxp files to PDF and JSON (text extraction).
Outputs go into the same folder as each source file.

Requirements: pip install pywin32
Tested against: QuarkXPress 2023/2024 on Windows
"""

import os
import sys
import json
import glob
import time
import logging
import win32com.client
import pythoncom
import argparse
```

```python
SOURCE_DIR   = r"E:\Callproject\assembled2\#out"
FILE_PATTERN = "**/*_9x.qxp"
QUARK_PROGID = "QuarkXPress.Application"   # COM ProgID – adjust if needed
QUARK_VISIBLE = True    # Set False for headless (less stable on some builds)
PAUSE_BETWEEN = 2.0     # Seconds to wait between files (give Quark breathing room)
CACHE_FILE = os.path.join(SOURCE_DIR, "_file_cache.json")

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s  %(levelname)-8s  %(message)s",
    handlers=[
        logging.StreamHandler(sys.stdout),
        logging.FileHandler(os.path.join(SOURCE_DIR, "_batch_export.log"), encoding="utf-8"),
    ],
)
log = logging.getLogger(__name__)
```

```python
def save_file_cache(file_list):
    with open(CACHE_FILE, "w", encoding="utf-8") as fh:
        json.dump(file_list, fh, indent=2)
    log.info("File list cached → %s  (%d files)", CACHE_FILE, len(file_list))

def load_file_cache():
    if os.path.exists(CACHE_FILE):
        with open(CACHE_FILE, "r", encoding="utf-8") as fh:
            file_list = json.load(fh)
        log.info("Loaded file list from cache  (%d files)", len(file_list))
        return file_list
    return None
```

```python
def get_quark_app():
    """Connect to a running QuarkXPress instance, or launch a new one."""
    pythoncom.CoInitialize()
    try:
        app = win32com.client.GetActiveObject(QUARK_PROGID)
        log.info("Attached to existing QuarkXPress instance.")
    except Exception:
        app = win32com.client.Dispatch(QUARK_PROGID)
        log.info("Launched new QuarkXPress instance.")
    app.Visible = QUARK_VISIBLE
    return app
```

```python
def open_document(app, qxp_path):
    """Open a QXP file and return the document object."""
    doc = app.OpenDocument(qxp_path)
    return doc
```

```python
def export_pdf(doc, pdf_path):
    """
    Export the open document to PDF.

    QuarkXPress exposes ExportLayout (QXP 9+) or PrintDocument depending on
    the version. Try ExportLayout first; fall back to the PDF print driver.
    """
    try:
        # QXP 9+ preferred method
        # ExportLayout(path, format)
        # format 1 = PDF  (constant: epfPDF or numeric 1)
        doc.ExportLayout(pdf_path, 1)
        log.info("    PDF exported via ExportLayout → %s", os.path.basename(pdf_path))
    except AttributeError:
        # Older COM interface fallback – uses the PDF print driver
        _export_pdf_via_print(doc, pdf_path)
```

```python
def _export_pdf_via_print(doc, pdf_path):
    """
    Fallback: drive the PDF export through QuarkXPress Print with PDF output.
    Uses the built-in 'Export as PDF' from the File menu equivalent.
    """
    try:
        ps = doc.PrintSetup          # PrintSetup object
        ps.Printer = "PDF"           # Must match your installed PDF printer name
        ps.OutputFile = pdf_path
        doc.Print()
        log.info("    PDF exported via PrintSetup → %s", os.path.basename(pdf_path))
    except Exception as e:
        log.warning("    PDF export fallback also failed: %s", e)
```

```python
def extract_text_to_dict(doc, qxp_path):
    """
    Walk every page → every box → collect text content.
    Returns a dict ready for json.dump().
    """
    result = {
        "source_file": qxp_path,
        "document_name": doc.Name,
        "page_count": doc.Pages.Count,
        "pages": [],
    }

    for page_idx in range(1, doc.Pages.Count + 1):
        page = doc.Pages.Item(page_idx)
        page_data = {
            "page_number": page_idx,
            "text_boxes": [],
        }

        for box_idx in range(1, page.Boxes.Count + 1):
            box = page.Boxes.Item(box_idx)

            # Only text boxes carry a .Text / .Contents property
            try:
                box_type = box.BoxType      # 1=Text, 2=Picture, 3=Empty, etc.
            except Exception:
                continue

            if box_type == 1:               # Text box
                try:
                    content = box.Text.PlainText   # Strip XPress tags → plain text
                except Exception:
                    try:
                        content = box.Contents     # Older API name
                    except Exception:
                        content = ""

                if content.strip():
                    page_data["text_boxes"].append({
                        "box_index": box_idx,
                        "text": content.strip(),
                    })

        result["pages"].append(page_data)

    return result
```

```python
def close_document(doc):
    """Close without saving."""
    try:
        doc.Close(False)    # False = don't save changes
    except Exception as e:
        log.warning("    Could not close document cleanly: %s", e)
```

```python
def main():
    parser = argparse.ArgumentParser()
    parser.add_argument(
        "--rescan",
        action="store_true",
        help="Ignore cached file list and re-scan the source directory.",
    )
    args = parser.parse_args()

    # ── File discovery (cached) ────────────────────────────────────────────────
    qxp_files = None if args.rescan else load_file_cache()

    if qxp_files is None:
        log.info("Scanning %s ...", SOURCE_DIR)
        pattern = os.path.join(SOURCE_DIR, FILE_PATTERN)
        qxp_files = sorted(glob.glob(pattern, recursive=True))
        save_file_cache(qxp_files)

    if not qxp_files:
        log.error("No *_9x.qxp files found under %s", SOURCE_DIR)
        sys.exit(1)

    log.info("Found %d QXP file(s) to process.", len(qxp_files))

    app = get_quark_app()

    success, failed = [], []

    for i, qxp_path in enumerate(qxp_files, start=1):
        stem = os.path.splitext(qxp_path)[0]          # path without extension
        pdf_path  = stem + ".pdf"
        json_path = stem + ".json"
        rel = os.path.relpath(qxp_path, SOURCE_DIR)

        log.info("[%d/%d] Processing: %s", i, len(qxp_files), rel)

        try:
            doc = open_document(app, qxp_path)

            # ── PDF export ──
            export_pdf(doc, pdf_path)

            # ── JSON / text extraction ──
            data = extract_text_to_dict(doc, qxp_path)
            with open(json_path, "w", encoding="utf-8") as fh:
                json.dump(data, fh, ensure_ascii=False, indent=2)
            log.info("    JSON saved → %s", os.path.basename(json_path))

            close_document(doc)
            success.append(rel)

        except Exception as exc:
            log.error("    FAILED: %s — %s", rel, exc)
            failed.append((rel, str(exc)))
            # Attempt to recover: close whatever is open and continue
            try:
                app.Documents.Item(1).Close(False)
            except Exception:
                pass

        time.sleep(PAUSE_BETWEEN)

    log.info("=" * 60)
    log.info("Done.  Success: %d   Failed: %d", len(success), len(failed))
    if failed:
        log.warning("Failed files:")
        for path, err in failed:
            log.warning("  %s  →  %s", path, err)
```

```python
if __name__ == "__main__":
    main()
```
