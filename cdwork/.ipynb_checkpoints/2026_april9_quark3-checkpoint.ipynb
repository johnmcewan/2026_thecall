```python
import subprocess
import os
import os.environ
from pathlib import Path
import pdfplumber
```

```python
def extract_quark_to_text_via_pdf(source_dir, target_dir, soffice_path):
    source_root = Path(source_dir).resolve()
    target_root = Path(target_dir).resolve()
    target_root.mkdir(parents=True, exist_ok=True)
    
    # Clean environment to prevent library conflicts
    clean_env = os.environ.copy()
    clean_env.pop("PYTHONHOME", None)
    clean_env.pop("PYTHONPATH", None)

    for file_path in source_root.rglob("*"):
        # Skip directories and existing PDFs/TXTs
        if file_path.is_dir() or file_path.suffix in [".pdf", ".txt"]:
            continue

        print(f"Processing: {file_path.name}")
        
        try:
            # 1. Convert Quark to PDF using LibreOffice (Draw engine)
            subprocess.run([
                soffice_path, "--headless",
                "--convert-to", "pdf",
                "--outdir", str(target_root),
                str(file_path)
            ], check=True, env=clean_env, capture_output=True)
            
            # 2. Path to the newly created PDF
            # Note: LibreOffice might add .pdf to the original filename
            pdf_path = target_root / f"{file_path.name}.pdf"
            txt_path = target_root / f"{file_path.name}.txt"

            if pdf_path.exists():
                # 3. Extract text from the PDF boxes
                with pdfplumber.open(pdf_path) as pdf:
                    # Join text from all pages
                    content = [page.extract_text() for page in pdf.pages if page.extract_text()]
                    full_text = "\n\n".join(content)
                
                with open(txt_path, "w", encoding="utf-8") as f:
                    f.write(full_text)
                
                # Cleanup intermediate PDF to save space
                pdf_path.unlink()
                print(f"   Successfully extracted to {txt_path.name}")

        except Exception as e:
            print(f"   Error processing {file_path.name}: {e}")
```
