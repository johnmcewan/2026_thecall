```python
import json

# Load both files
with open('final_metadata_summary.json', 'r') as f:
    master_list = json.load(f)

with open('nitearticles_v1.json', 'r') as f:
    raw_extraction = json.load(f)

# Create a big block of all text we have extracted so far to search against
all_extracted_text = " ".join([item.get('text_plain', '') for item in raw_extraction]).lower()

missing_articles = []

for article in master_list:
    title = article.get('Title', '').lower()
    # Use a unique-ish part of the title (first 20 chars) to see if it exists in our raw data
    search_snippet = title[:30] 
    
    if search_snippet not in all_extracted_text:
        missing_articles.append({
            'Title': article.get('Title'),
            'Date': article.get('pubdate'),
            'ProQuest_ID': article.get('StoreId')
        })

# Output the results
print(f"Found {len(missing_articles)} potential missing articles.")
for miss in missing_articles[:10]: # Show first 10
    print(f"- MISSING: {miss['Date']} | {miss['Title']}")
```
