```python
import winreg

def find_quark_in_registry():
    found = []
    # HKCR is where all COM ProgIDs live
    with winreg.OpenKey(winreg.HKEY_CLASSES_ROOT, "") as root:
        for i in range(winreg.QueryInfoKey(root)[0]):
            try:
                key_name = winreg.EnumKey(root, i)
                if "Quark" in key_name:
                    found.append(key_name)
            except OSError:
                break
    return found

print("Registered Quark-related names:")
for name in find_quark_in_registry():
    print(f" - {name}")
```

```python
# Usage
source = r"E:\Callproject\assembled2\#out_test\1_05-02 Call\Classifieds\Child Care"
output = r"E:\Callproject\assembled2\#out_test"
```

```python
def process_quark_files(source_dir, output_dir, pdf_style="Press Quality"):
    # Ensure output directory exists
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    try:
        # Initialize QuarkXPress
        print("Connecting to QuarkXPress...")
        qxp = win32com.client.Dispatch("QuarkXPress.Application")
        
        # Get all .qxp files
        files = [f for f in os.listdir(source_dir) if f.lower().endswith('.qxp')]
        
        for filename in files:
            file_path = os.path.join(source_dir, filename)
            base_name = os.path.splitext(filename)[0]
            print(f"Processing: {filename}")
            
            try:
                # Open Document: (Path, UseSettings, RemapFonts)
                doc = qxp.OpenDoc(file_path, 0, False)
                
                # 1. Export PDF
                pdf_path = os.path.join(output_dir, f"{base_name}.pdf")
                doc.ExportAsPDF(pdf_path, pdf_style)
                
                # 2. Export Text (Plain Text = 1, XPress Tags = 2)
                if doc.Stories.Count > 0:
                    txt_path = os.path.join(output_dir, f"{base_name}.txt")
                    doc.Stories(1).Save(txt_path, 1)
                
                # Close without saving (2 = DoNotSave)
                doc.Close(2)
                print(f"  - Successfully exported {filename}")
                
            except Exception as e:
                print(f"  - Error processing {filename}: {e}")
                
    except Exception as e:
        print(f"Failed to connect to QuarkXPress: {e}")
    finally:
        # Optional: Close Quark when finished
        # qxp.Quit()
        print("Batch processing complete.")

```

```python
process_quark_files(source, output)
```
