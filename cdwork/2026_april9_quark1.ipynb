```python
def extract_via_pdf_bridge(source_dir, target_dir, soffice_path):
    source_root = Path(source_dir).resolve()
    target_root = Path(target_dir).resolve()
    
    # Clean environment
    clean_env = os.environ.copy()
    clean_env.pop("PYTHONHOME", None)
    clean_env.pop("PYTHONPATH", None)

    for file_path in source_root.rglob("*"):
        if file_path.is_dir() or file_path.suffix == ".pdf":
            continue

        # 1. Convert to PDF
        print(f"Converting to PDF: {file_path.name}")
        try:
            subprocess.run([
                soffice_path, "--headless",
                "--convert-to", "pdf",
                "--outdir", str(target_root),
                str(file_path)
            ], check=True, env=clean_env)
            
            pdf_path = target_root / f"{file_path.name}.pdf"
            txt_path = target_root / f"{file_path.name}.txt"

            # 2. Extract text from the generated PDF
            if pdf_path.exists():
                with pdfplumber.open(pdf_path) as pdf:
                    full_text = "\n".join([page.extract_text() for page in pdf.pages if page.extract_text()])
                
                with open(txt_path, "w", encoding="utf-8") as f:
                    f.write(full_text)
                
                # Optional: Remove the intermediate PDF
                pdf_path.unlink()
                
        except Exception as e:
            print(f"Failed {file_path.name}: {e}")
```
