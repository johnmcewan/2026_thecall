```python
import threading, time
from pywinauto import Desktop
```

```python
def dismiss_quark_dialogs(stop_event):
    DISMISS = ["OK", "Repair it", "Continue", "Use Default"]
    print("Dialog watcher running...")
    while not stop_event.is_set():
        try:
            for win in Desktop(backend="win32").windows():
                try:
                    for child in win.children():
                        if child.window_text().strip() in DISMISS:
                            child.click()
                            print(f"  Auto-dismissed: '{child.window_text().strip()}'  [{win.window_text()}]")
                            time.sleep(0.5)
                except Exception:
                    pass
        except Exception:
            pass
        time.sleep(1)
    print("Dialog watcher stopped.")

stop_event = threading.Event()
threading.Thread(target=dismiss_quark_dialogs, args=(stop_event,), daemon=True).start()

# When the batch finishes, run: stop_event.set()
```
